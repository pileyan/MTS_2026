# Задание 1

На лекции в рамках разговора про Annoy разбирался пример про случайные прямые/плоскости. Представим, что у вас есть плоскость, на которую вы случайным образом бросаете прямые (как генерировать случайные прямые, мы обсуждали). Пересечение двух случайных прямых образует **узел**. Узлы, соединенные отрезками прямых образуют ребра. Ребра и узлы вместе представляют собой некий случайный граф. Вопрос: Как выглядит распределение степеней вершин этого графа?

Ответ получать так: многократно сгенерировать такой случайный граф (из одинакового достаточно большого количества брошенных прямых) и получить гистограммы распределения степеней вершин. 

По этим гистограммам посчитать базовые статистические характеристики: Среднее, медиана, дисперсия степеней
